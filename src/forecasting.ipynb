```python
import pandas as pd
df = pd.read_csv('../data/preprocessed_data/news_data_sentiment.csv')

daily_sentiment = (
    df.groupby(
        ["date", "company", "ticker"],
        as_index=False
    )
    .agg(
        sentiment_score=("sentiment_score", "mean"),
        positive_score=("positive_score", "mean"),
        negative_score=("negative_score", "mean"),
        neutral_score=("neutral_score", "mean"),
        n_articles=("nid", "nunique")
    )
)
```

```python
score_cols = [
    "positive_score",
    "negative_score",
    "neutral_score"
]

daily_sentiment["sentiment"] = (
    daily_sentiment[score_cols]
    .idxmax(axis=1)
    .str.replace("_score", "", regex=False)
)

daily_sentiment
```

```python
daily_sentiment.to_csv('../data/preprocessed_data/daily_sentiment.csv', index=False)
```
